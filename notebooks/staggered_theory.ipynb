```
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt

from sqw import build_walk_operator, run_walk_numpy, visualise_graph
```

## 1. A small weighted graph

```
G = nx.Graph()
for u, v, w in [('v1','v2',1), ('v2','v3',1), ('v2','v4',1), ('v3','v4',1)]:
    G.add_edge(u, v, weight=w)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
visualise_graph(G, after_clique_insertion=False, ax=axes[0])
res = build_walk_operator(G)
visualise_graph(res['H'], after_clique_insertion=True, ax=axes[1])
plt.tight_layout(); plt.show()
```

## 2. Build the walk operator
`build_walk_operator` performs clique insertion and returns `W`, the tessellation reflections, and the index maps.

```
res = build_walk_operator(G)
print('Hilbert-space dim (directed edges):', len(res['node_list']))
print('W is unitary:', np.allclose(res['W'].conj().T @ res['W'], np.eye(len(res['node_list']))))
```

## 3. Run the walk
Start from a uniform superposition over the directed edges of clique `v2` and track the clique probabilities.

```
ni = res['node_index']; nl = res['node_list']; cm = res['clique_map']
psi0 = np.zeros(len(nl), dtype=complex)
for e in [('v2','v1'), ('v2','v3'), ('v2','v4')]:
    psi0[ni[e]] = 1/np.sqrt(3)

final, traj = run_walk_numpy(res['W'], cm, ni, psi0, steps=30, plot_cliques=True)
```
